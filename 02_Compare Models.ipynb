In the previous example, we make a sample for lithium-ion battery in the DFN model. Since PyBaMM provides a couple more model, we will apply all of them in this notebook.
I observe there is no change with 1C discharge in all models 

```python
import pybamm
```

```python
Models = []
```

```python
Model_DFN = pybamm.lithium_ion.DFN()
Models.append(Model_DFN)

Model_BasicDFN = pybamm.lithium_ion.BasicDFN()
Models.append(Model_BasicDFN)

Model_SPMe = pybamm.lithium_ion.SPMe()
Models.append(Model_SPMe)

Model_SPM = pybamm.lithium_ion.SPM()
Models.append(Model_SPM)

print("Models to appended2: ", Models)
```

```python
#DFN MODEL RUN 1C

sims_DFN = pybamm.Simulation(Models[0])
sims_DFN.solve([0, 3600]) # 1 hour simulation
qp = pybamm.QuickPlot(sims_DFN.solution)
qp.plot (1)
```

```python
#BasicDFN MODEL RUN

sims_BasicDFN = pybamm.Simulation(Models[1])
sims_BasicDFN.solve([0, 3600]) # 1 hour simulation
qp = pybamm.QuickPlot(sims_BasicDFN.solution)
qp.plot (1)
```

```python
#SPM MODEL RUN

sims_SPM = pybamm.Simulation(Models[3])
sims_SPM.solve([0, 3600]) # 1 hour simulation
qp = pybamm.QuickPlot(sims_SPM.solution)
qp.plot (1)
```

```python
#DFN MODEL RUN

sims_SPMe = pybamm.Simulation(Models[2])
sims_SPMe.solve([0, 3600]) # 1 hour simulation
qp = pybamm.QuickPlot(sims_SPMe.solution)
qp.plot (1)
```

```python
parameter_values = pybamm.ParameterValues("Chen2020")
parameter_values
```

```python
#DFN MODEL RUN 5C
sims_DFN_5C = pybamm.Simulation(Models[0], parameter_values=parameter_values)
sims_DFN_5C.solve([0, 3600]) # 1 hour simulation
qp = pybamm.QuickPlot(sims_DFN_5C.solution)
qp.plot (1)
```
