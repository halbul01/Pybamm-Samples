We compared models and work with different parameters. We can also define charging/discharging protocol by using pybamm.experiment() API. 

Here there are some charging/discharging defines. 
"Discharge at 1C for 0.5 hours",
"Discharge at C/20 for 0.5 hours",
"Charge at 0.5 C for 45 minutes",
"Discharge at 1 A for 90 seconds",
"Charge at 200mA for 45 minutes",
"Discharge at 1 W for 0.5 hours",
"Charge at 200 mW for 45 minutes",
"Rest for 10 minutes",
"Hold at 1 V for 20 seconds",
"Charge at 1 C until 4.1V",
"Hold at 4.1 V until 50 mA",
"Hold at 3V until C/50",

```python
import pybamm
```

```python
first_experiment = pybamm.Experiment([
    "Discharge at 2C for 0.5 hours",
    'Rest for 0.5 hours',
    "Charge at 1C for 0.5 hours",
    'Rest for 0.5 hours'
])
```

```python
model = pybamm.lithium_ion.DFN() #DFN model
sim = pybamm.Simulation(model, experiment=first_experiment) #Simulation of the model with the experiment
sim.solve() #Solving the simulation
```

```python
#sim.plot() #Plotting the results
```

```python
qp = pybamm.QuickPlot(sim.solution)
qp.plot(1) #Plotting the results
```
