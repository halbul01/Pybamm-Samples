This is the first simulation of the PyBaMM. PyBaMM is the an open source tool that simulate whole battery and its simulation in Python. I observerd a discharge of lithium-ion battery here.

Here you can see my first simulation which run Lithium-ion battery as DFN (Doyle-Fuller-Newman) model. I can observe here 1C discharge at 1 hour.



```python
import pybamm
```

```python
model = pybamm.lithium_ion.DFN()
```

```python
sim = pybamm.Simulation(model)
```

```python
sim.solve([0, 3600]) # 1 hour simulation
```

```python
sim.plot()
```
