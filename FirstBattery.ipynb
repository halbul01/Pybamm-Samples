```python
import pybamm
```

```python
model = pybamm.lithium_ion.DFN()
```

```python
sim = pybamm.Simulation(model)
```

```python
sim.solve([0, 3600])
```

```python
sim.plot()
```

Statik grafik (GitHub'da görünür): t = 0.5 saatteki durum

```python
qp = pybamm.QuickPlot(sim.solution)
qp.plot(0.5)  # 0.5. saatteki durum
```
