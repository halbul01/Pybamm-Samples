In this example, I will set some parameters which remained in last exercise. Current and Capacity

```python
import pybamm
```

```python
parameter_values = pybamm.ParameterValues("Chen2020") #Importing parameter values from the parameter set
```

```python
parameter_values #Viewing the parameter values
```

```python
#'Nominal cell capacity [A.h]': 5.0 which means 1C = 5A. First let's see 1C plottings.
model = pybamm.lithium_ion.DFN() #DFN model
sim = pybamm.Simulation(model, parameter_values=parameter_values) #Simulation of the model with the parameter values
sim.solve([0, 3600]) # 1 hour simulation
qp = pybamm.QuickPlot(sim.solution)
qp.plot(1) #Plotting the results
```
