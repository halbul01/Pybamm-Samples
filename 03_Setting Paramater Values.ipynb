In this example, I will set some parameters which remained in last exercise. Current and Capacity

```python
import pybamm
```

```python
parameter_values = pybamm.ParameterValues("Chen2020") #Importing parameter values from the parameter set
```

```python
parameter_values #Viewing the parameter values
```

```python
#'Nominal cell capacity [A.h]': 5.0 which means 1C = 5A. First let's see 1C plottings.
model = pybamm.lithium_ion.DFN() #DFN model
sim_1c = pybamm.Simulation(model, parameter_values=parameter_values) #Simulation of the model with the parameter values
sim_1c.solve([0, 3600]) # 1 hour simulation
qp = pybamm.QuickPlot(sim_1c.solution)
qp.plot(1) #Plotting the results
```

```python
#Now let's see 2C plottings. 2C means 10A for this cell.
parameter_values["Current function [A]"] = 10 #Changing the current to 10A for 2C
sim_2C = pybamm.Simulation(model, parameter_values=parameter_values) #Simulation of the model with the parameter values
sim_2C.solve([0, 1800]) # 30 minutes simulation
qp = pybamm.QuickPlot(sim_2C.solution)
qp.plot(1) #Plotting the results
```
